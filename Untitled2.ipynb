```python
import re
import json
import logging
from datetime import datetime
import requests

# 1. Configuración de integracion.log
LOG_FILE = "integracion.log"
logging.basicConfig(
    filename=LOG_FILE,
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S"
)

def log_peticion(metodo: str, url: str, status_code: int, detalle: str = ""):
    """Registra cada operación en el archivo integracion.log."""
    mensaje = f"{metodo.upper()} {url} -> Status: {status_code} | {detalle}"
    logging.info(mensaje)

def mostrar(titulo: str, respuesta: requests.Response):
    """Muestra en pantalla el resumen de la petición y respuesta HTTP."""
    print(f"\n{'='*70}")
    print(f" OPERACIÓN: {titulo}")
    print(f"{'='*70}")
    print(f"Método: {respuesta.request.method}")
    print(f"URL:    {respuesta.request.url}")
    print(f"Estado: {respuesta.status_code} {respuesta.reason}")
    try:
        data = respuesta.json()
        print(f"Respuesta JSON:\n{json.dumps(data, indent=2)}")
    except ValueError:
        print("Cuerpo vacío o sin formato JSON")
    print(f"{'-'*70}")

def validar_datos_cliente(datos: dict) -> tuple[bool, str]:
    """
    Firma de validación previa:
    - Nombre no vacío.
    - Dominio corporativo obligatorio (@techdistribuidora.ec).
    - Género: male | female.
    - Estado: active | inactive.
    """
    campos_obligatorios = ["name", "email", "gender", "status"]
    for campo in campos_obligatorios:
        if campo not in datos or not str(datos[campo]).strip():
            return False, f"Campo obligatorio faltante o vacío: '{campo}'"

    # Validación estricta de correo corporativo
    patron_email = r"^[\w\.-]+@techdistribuidora\.ec$"
    if not re.match(patron_email, datos["email"]):
        return False, f"El correo '{datos['email']}' no cumple con el dominio institucional @techdistribuidora.ec"

    if datos["gender"] not in ["male", "female"]:
        return False, f"Género inválido: {datos['gender']}. Valores permitidos: 'male', 'female'"

    if datos["status"] not in ["active", "inactive"]:
        return False, f"Estado inválido: {datos['status']}. Valores permitidos: 'active', 'inactive'"

    return True, "Validación exitosa"

def ejecutar_peticion(metodo: str, url: str, headers: dict = None, json_data: dict = None, params: dict = None) -> requests.Response:
    """Ejecuta la petición, gestiona códigos de error conocidos y escribe en el log."""
    try:
        resp = requests.request(
            method=metodo,
            url=url,
            headers=headers,
            json=json_data,
            params=params,
            timeout=10
        )

        # Registro en integracion.log
        log_peticion(metodo, url, resp.status_code)

        # Manejo de respuestas y control de fallos
        if resp.status_code == 401:
            print(f"[ALERTA 401] Autenticación rechazada. Token ausente o caducado en {url}")
        elif resp.status_code == 404:
            print(f"[ALERTA 404] Recurso no encontrado en {url}")
        elif resp.status_code == 422:
            print(f"[ALERTA 422] Entidad no procesable: La API rechazó los datos (ej. email duplicado).")
        elif resp.status_code == 429:
            print(f"[ALERTA 429] Límite de tasa excedido (Rate Limit). Espere antes de reintentar.")

        return resp
    except requests.RequestException as e:
        logging.error(f"Error de conexión en {metodo} {url}: {str(e)}")
        print(f"[ERROR DE RED] No se pudo completar la conexión: {e}")
        return None
```

```python
URL_SANDBOX = "https://jsonplaceholder.typicode.com/posts"

print(">>> INICIANDO PRUEBAS EN SANDBOX (JSONPlaceholder)...")

# 1. GET lista
r_get = ejecutar_peticion("GET", URL_SANDBOX, params={"_limit": 2})
mostrar("GET - Consultar Posts", r_get)

# 2. POST nuevo
nuevo_post = {"title": "Integración TechDistribuidora", "body": "Prueba inicial", "userId": 1}
r_post = ejecutar_peticion("POST", URL_SANDBOX, json_data=nuevo_post)
mostrar("POST - Crear Post simulado", r_post)
id_sandbox = r_post.json().get("id")

# 3. PUT reemplazo
put_post = {"id": 1, "title": "Título Totalmente Modificado", "body": "Cuerpo reemplazado", "userId": 1}
r_put = ejecutar_peticion("PUT", f"{URL_SANDBOX}/1", json_data=put_post)
mostrar("PUT - Reemplazo completo Post #1", r_put)

# 4. PATCH parcial
patch_post = {"title": "Solo Título Cambiado"}
r_patch = ejecutar_peticion("PATCH", f"{URL_SANDBOX}/1", json_data=patch_post)
mostrar("PATCH - Actualización parcial Post #1", r_patch)

# 5. GET de comprobación de persistencia
r_check = ejecutar_peticion("GET", f"{URL_SANDBOX}/1")
mostrar("GET - Verificación de persistencia en Post #1", r_check)
print("Observación: El Post #1 mantiene su contenido original. JSONPlaceholder no persiste los cambios.")
```

```python
import getpass

print("Ingrese su Access Token de GoRest (obtenido en gorest.co.in):")
GOREST_TOKEN = getpass.getpass("Token: ")

BASE_URL = "https://gorest.co.in/public/v2"
HEADERS = {
    "Authorization": f"Bearer {GOREST_TOKEN.strip()}",
    "Content-Type": "application/json",
    "Accept": "application/json"
}

if not GOREST_TOKEN:
    print("Advertencia: No ingresó ningún token.")
else:
    print("Token cargado en memoria de forma segura.")
```

```python
import time

usuario_id = None
# Generar sufijo único usando timestamp para evitar duplicados en el email
timestamp = int(datetime.now().timestamp())

# --- 1. GET con paginación (5 clientes) ---
r_list = ejecutar_peticion("GET", f"{BASE_URL}/users", headers=HEADERS, params={"page": 1, "per_page": 5})
mostrar("GET - Lista de Clientes Paginada (200 OK)", r_list)

# --- 2. POST Registrar Cliente ---
nuevo_cliente = {
    "name": "Santiago Distribuidor",
    "email": f"santiago.tech.{timestamp}@techdistribuidora.ec",
    "gender": "male",
    "status": "active"
}

valido, msg = validar_datos_cliente(nuevo_cliente)
if valido:
    r_create = ejecutar_peticion("POST", f"{BASE_URL}/users", headers=HEADERS, json_data=nuevo_cliente)
    mostrar("POST - Registro de Cliente (201 Created)", r_create)
    if r_create.status_code == 201:
        usuario_id = r_create.json().get("id")
else:
    print(f"Rechazado por validación local: {msg}")

# --- 3. GET Verificar Registro ---
if usuario_id:
    r_get_id = ejecutar_peticion("GET", f"{BASE_URL}/users/{usuario_id}", headers=HEADERS)
    mostrar("GET - Verificar Registro Creado (200 OK)", r_get_id)

    # --- 4. PUT Reemplazar Datos Completos ---
    cliente_modificado = {
        "name": "Santiago Conlago Actualizado",
        "email": f"santiago.modificado.{timestamp}@techdistribuidora.ec",
        "gender": "male",
        "status": "active"
    }
    valido_put, msg_put = validar_datos_cliente(cliente_modificado)
    if valido_put:
        r_put = ejecutar_peticion("PUT", f"{BASE_URL}/users/{usuario_id}", headers=HEADERS, json_data=cliente_modificado)
        mostrar("PUT - Reemplazo Completo de Datos (200 OK)", r_put)

    # --- 5. PATCH Cambiar Solo Estado ---
    cliente_patch = {"status": "inactive"}
    r_patch = ejecutar_peticion("PATCH", f"{BASE_URL}/users/{usuario_id}", headers=HEADERS, json_data=cliente_patch)
    mostrar("PATCH - Desactivación Parcial de Cliente (200 OK)", r_patch)

    # --- 6. AMPLIACIÓN: Recurso Anidado (/todos) ---
    nueva_tarea = {
        "title": "Verificar contrato comercial TechDistribuidora",
        "due_on": "2026-11-15T00:00:00.000+00:00",
        "status": "pending"
    }
    r_todo_create = ejecutar_peticion("POST", f"{BASE_URL}/users/{usuario_id}/todos", headers=HEADERS, json_data=nueva_tarea)
    mostrar("POST - Crear Tarea Anidada (201 Created)", r_todo_create)

    if r_todo_create.status_code == 201:
        todo_id = r_todo_create.json().get("id")
        # Marcar tarea como completada vía PATCH
        patch_todo = {"status": "completed"}
        r_todo_patch = ejecutar_peticion("PATCH", f"{BASE_URL}/todos/{todo_id}", headers=HEADERS, json_data=patch_todo)
        mostrar("PATCH - Completar Tarea Anidada (200 OK)", r_todo_patch)

    # --- 7. DELETE Dar de baja al Cliente ---
    r_del = ejecutar_peticion("DELETE", f"{BASE_URL}/users/{usuario_id}", headers=HEADERS)
    mostrar("DELETE - Dar de baja al Cliente (204 No Content)", r_del)

    # --- 8. GET Comprobar Baja (Debe responder 404) ---
    r_check_deleted = ejecutar_peticion("GET", f"{BASE_URL}/users/{usuario_id}", headers=HEADERS)
    mostrar("GET - Comprobación Post-Baja (404 Not Found)", r_check_deleted)
```

```python
print("\n>>> EJECUTANDO PRUEBAS DE CAMINO DE ERROR...")

# Prueba 1: 401 Unauthorized (Token erróneo o falsificado)
headers_invalidos = {"Authorization": "Bearer TOKEN_INVALIDO_XYZ", "Content-Type": "application/json"}
r_err_401 = ejecutar_peticion("GET", f"{BASE_URL}/users", headers=headers_invalidos)
mostrar("PRUEBA ERROR: 401 Unauthorized", r_err_401)

# Prueba 2: 422 Unprocessable Entity (Envío intencional de email no válido directo a la API)
payload_invalido = {
    "name": "Usuario Prueba Error",
    "email": "correo_sin_formato_correcto",
    "gender": "male",
    "status": "active"
}
r_err_422 = ejecutar_peticion("POST", f"{BASE_URL}/users", headers=HEADERS, json_data=payload_invalido)
mostrar("PRUEBA ERROR: 422 Unprocessable Entity", r_err_422)
```
